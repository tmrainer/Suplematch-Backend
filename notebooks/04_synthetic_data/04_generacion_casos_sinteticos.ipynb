# Casos sintéticos

Genera un dataset temporal desde reglas curadas.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd()
if (ROOT / 'Suplematch-Backend').exists():
    ROOT = ROOT / 'Suplematch-Backend'
while ROOT.name != 'Suplematch-Backend' and ROOT.parent != ROOT:
    ROOT = ROOT.parent

sys.path.insert(0, str(ROOT))
from scripts.training.entrenar_modelo_condiciones import load_knowledge, generate_dataset, dataset_eda
ROOT
```

## Generación

```python
ROWS = 500
SEED = 42
knowledge = load_knowledge()
df = generate_dataset(knowledge, n_rows=ROWS, seed=SEED)
df.head(10)
```

## EDA

```python
report = dataset_eda(df, knowledge['conditions']['labels'])
label_rows = [{'condition': key, **value} for key, value in report['label_prevalence'].items()]
display(report['summary'])
pd.DataFrame(label_rows).sort_values('positive_rate', ascending=False).head(20)
```

## Exportación temporal

```python
RUN_EXPORT = False
out = ROOT / 'var/notebook_synthetic_condition_cases.csv'
if RUN_EXPORT:
    out.parent.mkdir(exist_ok=True)
    df.to_csv(out, index=False)
{'path': str(out.relative_to(ROOT)), 'written': RUN_EXPORT, 'rows': len(df)}
```
