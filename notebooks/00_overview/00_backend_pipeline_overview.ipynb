# Overview backend

Mapa ejecutable de carpetas, scripts y artefactos principales.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
if (ROOT / 'Suplematch-Backend').exists():
    ROOT = ROOT / 'Suplematch-Backend'
while ROOT.name != 'Suplematch-Backend' and ROOT.parent != ROOT:
    ROOT = ROOT.parent

assert (ROOT / 'app').exists()
assert (ROOT / 'scripts').exists()
assert (ROOT / 'data').exists()
ROOT
```

## Inventario

```python
items = []
for path in ['app/domains', 'app/ml/runtime', 'data/knowledge', 'data/catalog', 'scripts/training', 'scripts/scraping', 'models/runtime', 'tests']:
    target = ROOT / path
    items.append({'path': path, 'exists': target.exists(), 'files': len(list(target.rglob('*'))) if target.exists() else 0})
pd.DataFrame(items)
```
