# Evaluación

Ejecuta golden cases, holdout y benchmark cuando `RUN` sea `True`.

```python
from pathlib import Path
import subprocess
import pandas as pd

ROOT = Path.cwd()
if (ROOT / 'Suplematch-Backend').exists():
    ROOT = ROOT / 'Suplematch-Backend'
while ROOT.name != 'Suplematch-Backend' and ROOT.parent != ROOT:
    ROOT = ROOT.parent

scripts = {
    'golden': ROOT / 'scripts/training/evaluar_casos_golden_condiciones.py',
    'holdout': ROOT / 'scripts/training/evaluar_holdout_condiciones.py',
    'suite': ROOT / 'scripts/training/ejecutar_suite_benchmark_condiciones.py',
    'quality': ROOT / 'scripts/validate_condition_model_quality.py',
}
assert all(path.exists() for path in scripts.values())
ROOT
```

## Comandos

```python
commands = {
    'golden': ['python', str(scripts['golden'])],
    'holdout': ['python', str(scripts['holdout']), '--rows', '1000'],
    'suite': ['python', str(scripts['suite']), '--skip-train', '--limit-per-cycle', '500'],
    'quality': ['python', str(scripts['quality'])],
}
pd.DataFrame([{'name': key, 'command': ' '.join(value)} for key, value in commands.items()])
```

## Ejecución

```python
RUN = False
selected = ['golden']
outputs = []
if RUN:
    for name in selected:
        completed = subprocess.run(commands[name], cwd=ROOT, text=True, capture_output=True, check=True)
        outputs.append({'name': name, 'stdout': completed.stdout[-2000:]})
outputs
```

## Reportes

```python
report_dir = ROOT / 'data/reports/condition_model'
reports = sorted(report_dir.glob('*.csv')) if report_dir.exists() else []
pd.DataFrame([{'file': path.name, 'size': path.stat().st_size} for path in reports]).sort_values('file')
```
